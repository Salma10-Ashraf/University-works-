```python
import pandas as pd 
```

```python
df=pd.read_csv("iris.csv")
df
```

```python
x=df.drop(["Species"],axis=1)
x
```

```python
y=df["Species"]
y
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,train_size=0.8,random_state=42)
```

knn

```python
from sklearn.neighbors import KNeighborsClassifier
```

```python
knn=KNeighborsClassifier()
```

```python
knn.fit(x_train,y_train)
```

```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score
```

```python
train_pred=knn.predict(x_train)
train_acc= accuracy_score(y_train,train_pred)
print('accuracy_score',train_acc)
```

```python
test_pred=knn.predict(x_test)
test_acc=accuracy_score(y_test,test_pred)       
print('accuracy_score',test_acc)
```

```python
train_pre = precision_score(y_train, train_pred, average='macro')
print('precision_score', train_pre)

train_rec = recall_score(y_train, train_pred, average='macro')
print('recall_score', train_rec)

train_f1 = f1_score(y_train, train_pred, average='macro')
print('f1_score', train_f1)
```

```python
test_pre = precision_score(y_test, test_pred, average='macro')
print('precision_score', test_pre)

test_rec = recall_score(y_test, test_pred, average='macro')
print('recall_score', test_rec)

test_f1 = f1_score(y_test, test_pred, average='macro')
print('f1_score', test_f1)
```
