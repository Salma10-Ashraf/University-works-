```python
from itertools import zip_longest
```

```python
Result=requests.get('https://wuzzuf.net/search/jobs?q=Data%20Anlysis&start=0')
```

```python
src=Result.content
```

```python
print(src)
```

```python
soup=BeautifulSoup(src,'lxml')
print(soup)
```

```python
job_titles=soup.find_all("h2",{"class":"css-193uk2c"})
print(job_titles)
```

```python
companys_name=soup.find_all('a',{'class','css-ipsyv7'})
print(companys_name)
```

```python
locations_name=soup.find_all('span',{'class','css-16x61xq'})
print(locations_name)
```

```python
job_skills=soup.find_all('div',{'class','css-1rhj4yg'})
print(job_skills)
```

```python
job_title=[]
company_name=[]
location_name=[]
job_skill=[]
links=[]
```

```python
for i in range(len(job_titles)):
    job_title.append(job_titles[i].text)
    links.append(f"https://wuzzuf.net{job_titles[i].find('a').attrs['href']}")
    company_name.append(companys_name[i].text)
    location_name.append(locations_name[i].text)



    current_skills = job_skills[i].text.strip()
    
    unwanted = ["Remote", "Full Time", "On-site", "Part Time", "Work from home","Freelance"]
    
    for word in unwanted:
        current_skills = current_skills.replace(word, "")
    
    
    clean_skills = current_skills.strip().strip('').strip()
    
    job_skill.append(clean_skills)
  
    
print(job_title)
print("-----------------------------")
print("-----------------------------")
print(company_name)
print("-----------------------------")
print("-----------------------------")
print(location_name)
print("-----------------------------")
print("-----------------------------")
print(job_skill)
print("-----------------------------")
print("-----------------------------")
print(links)
print("-----------------------------")
print("-----------------------------")
```

```python
file_list=[job_title,company_name,location_name,job_skill,links]
exported=zip_longest(*file_list)
```

```python
with open(r"C:\Users\Electronica Care\Documents\Data_Anlysis_Job.csv",'w') as Data_Anlysis_Job:
    wr=csv.writer(Data_Anlysis_Job)
    wr.writerow(['job_titles','companys_name','locations','job_skills','links'])
    wr.writerows(exported)
```

```python
import pandas as pd

df1 = pd.read_csv(r"C:\Users\Electronica Care\Documents\Data_Anlysis_Job.csv",encoding='latin1')
df2 = pd.read_csv(r"C:\Users\Electronica Care\Documents\web_data_anlysis_2.csv",encoding='latin1')
df1.head()
```

```python
df2.head()
```

```python
df = pd.concat([df1, df2], ignore_index=True)
df
```

```python
df.info()
```

```python
df.shape
```

```python
df.rename(columns={'companys_name':'company_name'}, inplace=True)
df['company_name']=df['company_name'].str.strip()
df['company_name'] = df['company_name'].str.replace('-', '')
df['job_titles']=df['job_titles'].str.lower()
df['job_titles'] = df['job_titles'].str.title()
df['locations'] = df['locations'].str.title()
```

```python
df.isnull().sum()
```

```python
import numpy as np
for _ in range(10):
    row = np.random.randint(0, len(df))
    col = np.random.randint(0, len(df.columns))
    df.iat[row, col] = np.nan

df.head(15)
```

```python
df.isnull().sum()
```

```python
df["job_titles"].fillna(df["job_titles"].mode()[0], inplace=True)
df["company_name"].fillna(df["company_name"].mode()[0], inplace=True)
df["locations"].fillna(df["locations"].mode()[0], inplace=True)
df["job_skills"].fillna(df["job_skills"].mode()[0], inplace=True)
df["links"].fillna(df["links"].mode()[0],inplace=True)
df
```

```python
df.isnull().sum()
```

```python
duplicates = df.sample(n=5)
df = pd.concat([df, duplicates], ignore_index=True)
df.head(15)
```

```python
df.duplicated().sum()
```

```python
df.drop_duplicates(inplace=True)
df.head()
```

```python
df.duplicated().sum()
```

```python
df.to_csv(r"C:\Users\Electronica Care\Documents\cleaning_data_anlaysis.csv", index=False)
```

```python
import pandas as pd
import sqlite3
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv(r"C:\Users\Electronica Care\Documents\cleaning_data_anlaysis.csv",encoding='latin1')
df.head()
```

```python
conn = sqlite3.connect("jobs.db")
df.to_sql("jobs", conn, if_exists="replace", index=False)
print("data saved to database")
```

```python
query1 = pd.read_sql("SELECT * FROM jobs LIMIT 10", conn)
query1
```

```python
query2 = pd.read_sql("SELECT COUNT(*) as total_jobs FROM jobs", conn)
query2
```

```python
query3 = pd.read_sql("SELECT job_titles, COUNT(*) as count FROM jobs GROUP BY job_titles ORDER BY count DESC", conn)
query3
```

```python
query4 = pd.read_sql("SELECT company_name, COUNT(*) as count FROM jobs GROUP BY company_name ORDER BY count DESC LIMIT 10", conn)
query4
```

```python
query5 = pd.read_sql("SELECT locations, COUNT(*) as count FROM jobs GROUP BY locations ORDER BY count DESC", conn)
query5
```

```python
query6 = pd.read_sql("SELECT job_titles, COUNT(*) as count FROM jobs GROUP BY job_titles HAVING count > 3", conn)
query6
```

```python
query7 = pd.read_sql("SELECT * FROM jobs WHERE locations LIKE '%Cairo%'", conn)
query7
```

```python
query8 = pd.read_sql("SELECT * FROM jobs WHERE job_titles = 'Data Analyst'", conn)
query8
```

```python
query9 = pd.read_sql("SELECT job_titles, locations, COUNT(*) as count FROM jobs GROUP BY job_titles, locations ORDER BY count DESC LIMIT 15", conn)
query9
```

```python
query10 = pd.read_sql("SELECT company_name, COUNT(*) as jobs_count FROM jobs GROUP BY company_name HAVING jobs_count >= 2 ORDER BY jobs_count DESC", conn)
query10
```

```python
query11 = pd.read_sql("SELECT DISTINCT job_titles FROM jobs ORDER BY job_titles", conn)
query11
```

```python
query12 = pd.read_sql("SELECT locations, COUNT(DISTINCT company_name) as num_companies FROM jobs GROUP BY locations ORDER BY num_companies DESC", conn)
query12
```
