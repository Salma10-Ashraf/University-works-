```python
import pandas as pd                             #import libarary
import numpy as np
from sklearn.model_selection import train_test_split
```

```python
df=pd.read_csv("titanic.csv")   #load data set
```

```python
df.head()     #show first 5 rows
```

```python
df.shape              #no of rows and columns
```

```python
df.duplicated().sum()    #no of duplicate records
```

```python
df = df.drop_duplicates()      #remove duplicate records but this data not contain dublicate 
df
```

```python
df.isnull().sum()
```

```python
df[df['Age'].isnull()].index                  # #what index is missing in Age
```

```python
df['Age']=df['Age'].fillna(df['Age'].mean())    #missing..numerical..mean
df['Age'].head(7)                    #index[5] missing
```

```python
df[df['Embarked'].isnull()].index                #what index is missing in Embarked
```

```python
df['Embarked']=df['Embarked'].fillna(df['Embarked'].mode()[0])     # #missing..string..mode
df['Embarked'].head(63)                     #index[61]  missing 
```

```python
df[df['Cabin'].isnull()].index                #high missing data so we remove
```

```python
if 'Cabin' in df.columns:
    df.drop('Cabin', axis=1, inplace=True)
```

```python
df
```

```python
df.drop(['PassengerId', 'Name', 'Ticket'], axis=1, inplace=True)   #remove irrelevant data
df
```

```python
(df['Age'] > 0) & (df['Age'] < 80)
```

```python
df = df[(df['Age'] > 0) & (df['Age'] < 80)]        #remove noise
df
```

```python
corr_matrix = df.corr(numeric_only=True)          #calculate correlation
```

```python
to_drop = set()                                  # empty set to store columns we delete
```

```python
for i in range(len(corr_matrix.columns)):
    for j in range(i):
        if abs(corr_matrix.iloc[i, j]) >= 0.8:
            colname = corr_matrix.columns[i]
            to_drop.add(colname)

df.drop(columns=to_drop, inplace=True)
df.corr(numeric_only=True)
```

```python
df['Age_group'] = pd.cut(df['Age'],                
                         bins=[0, 12, 20, 40, 60, 80],
                         labels=['Child', 'Teen', 'Adult', 'Mid-age', 'Old'])

df.drop('Age', axis=1)                #discretization
df
```

```python
df = pd.get_dummies(df, drop_first=True)           #convert cat to string
```

```python
train_df, test_df = train_test_split(df, test_size=0.3, random_state=42)            #split dataset
```

```python
train_df.to_csv("train_data.csv", index=False)          
test_df.to_csv("test_data.csv", index=False)

print("Data preprocessing completed successfully")
```
