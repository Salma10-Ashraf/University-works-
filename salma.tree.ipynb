```python
import pandas as pd #1 import
```

```python
data=pd.read_csv('iris.csv')  #2 read
data.head()
```

```python
x=data.drop('Species',axis=1)  #3 seprate feature from data
x
```

```python
y=data['Species']  
y
```

```python
y.value_counts()  #4  transform catogries label into int
```

```python
from sklearn.model_selection import train_test_split   
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,train_size=0.8)  #5 split to train and test
```

```python
x_train.head()
```

```python
x_test.head()
```

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn import tree
```

```python
tree_model=DecisionTreeClassifier(criterion='entropy') #6 create object
```

```python
tree_model.fit(x_train,y_train)  #7 fit/train the training data
```

```python
from sklearn.metrics import accuracy_score
```

```python
train_pred=tree_model.predict(x_train)  #6
train_acc=accuracy_score(y_train,train_pred)  #9
print('train accuracy =',train_acc)
```

```python
test_pred=tree_model.predict(x_test)
test_acc=accuracy_score(y_test,test_pred)  #10
print('train accuracy =',test_acc) 
```

```python
import matplotlib.pyplot as plt
```

```python
tree.plot_tree(tree_model,
               feature_names=x.columns,
               filled=True,
              class_names=['setosa','versicolor','virginplot'])
plt.show()
```
