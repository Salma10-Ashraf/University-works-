```python
#5.Write a program to check whether a number is prime using a 
#loop and condition. 


no = int(input("Enter a number: "))
is_prime = True

if no <= 1:
    is_prime = False
else:
    for i in range(2, no):
        if no % i == 0:
            is_prime = False
            break

if is_prime:
    print("Prime number")
else:
    print("Not prime")
```

```python
#6. Reverse a given integer using a loop. 

no = int(input("Enter a number: "))
reverse = 0

while no > 0:
    digit = no % 10
    reverse = reverse * 10 + digit
    no //= 10

print("Reversed number =", reverse)
```

```python
#7. Count the number of vowels in a given string using a loop and 
#condition. 

text = input("Enter a string: ")
count = 0

for ch in text:
    if ch.lower() in "aeiou":
        count += 1

print("Number of vowels =", count)
```

```python
#8. Generate a random number between 1 and 20. Keep asking the 
#user to guess until they get it correct. 

import random

number = random.randint(1, 20)

while True:
    guess = int(input("Guess the number (1-20): "))

    if guess == number:
        print("Correct! 🎉")
        break
    elif guess < number:
        print("Too small!")
    else:
        print("Too big!")
```

```python
#10. Write a function that prints the first n Fibonacci numbers 
#using a loop.

def fibonacci(n):
    a = 0
    b = 1
    
    for i in range(n):
        print(a)
        a, b = b, a + b

fibonacci(10)
```

```python
#11. Create a function that checks whether a given string is a 
#palindrome. 

def is_palindrome(text):
    if text == text[::-1]:
        return True
    else:
        return False

print(is_palindrome("madam"))
```

```python
#12. Write a function that takes a list of numbers and returns 
#the count of even and odd numbers. 

def count_even_odd(numbers):
    even = 0
    odd = 0
    
    for num in numbers:
        if num % 2 == 0:
            even += 1
        else:
            odd += 1
            
    return even, odd

print(count_even_odd([1,2,3,4,5,6]))
```

```python
#13. Write a function that takes a student's score and returns: 
#• A (90–100) 
#• B (80–89) 
#• C (70–79) 
#• D (60–69) 
#• F (below 60) 

def grade(score):
    if 90 <= score <= 100:
        return "A"
    elif 80 <= score <= 89:
        return "B"
    elif 70 <= score <= 79:
        return "C"
    elif 60 <= score <= 69:
        return "D"
    else:
        return "F"

print(grade(85))
```
