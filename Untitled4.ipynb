```python
import numpy as np
import matplotlib.pyplot as plt

data = np.array([
 [1.1, 39343],
 [1.5, 46205],
 [2.0, 37731],
 [2.2, 43525],
 [2.9, 39891],
 [3.0, 56642],
 [3.2, 60150],
 [3.8, 54445],
 [4.0, 56957],
 [4.5, 61111],
 [5.0, 67938],
 [5.3, 66029],
 [5.9, 83088],
 [6.0, 81363],
 [6.8, 93940],
 [7.1, 91738],
 [7.9, 98273],
 [8.2, 101302],
 [8.7, 113812],
 [9.0, 109431]
])

X = data[:, 0]
y = data[:, 1]
m = len(y)

X = (X - np.mean(X)) / np.std(X)
y = (y - np.mean(y)) / np.std(y)

X = X.reshape(m, 1)
y = y.reshape(m, 1)

theta0 = 0
theta1 = 0
alpha = 0.01
iterations = 1000

costs = []

for i in range(iterations):
    y_pred = theta0 + theta1 * X
    error = y_pred - y
    cost = (1/(2*m)) * np.sum(error ** 2) 
    costs.append(cost)
    theta0 = theta0 - alpha * (1/m) * np.sum(error)
    theta1 = theta1 - alpha * (1/m) * np.sum(error * X)

print(theta0, theta1, costs[-1])

plt.plot(range(iterations), costs)
plt.xlabel('Iterations')
plt.ylabel('Cost')
plt.title('Cost Function vs Iterations')
plt.show()

plt.scatter(X, y, color='red')
plt.plot(X, theta0 + theta1*X, color='blue')
plt.show()
```

```python
*How Learning Rate Affects Convergence*

The learning rate controls how fast gradient descent learns.
When the learning rate is too small, the cost goes down very slowly and the model needs many iterations to converge.
When the learning rate is too large, the cost may increase or jump up and down, and the algorithm does not converge.
A medium value makes the cost decrease smoothly and reach the minimum faster.
From the experiment  a learning rate like 0.1 worked well after normalization.
```

```python
import numpy as np
import matplotlib.pyplot as plt

data = np.array([
 [35, 0],
 [45, 0],
 [52, 0],
 [60, 0],
 [62, 0],
 [65, 0],
 [68, 0],
 [70, 1],
 [72, 1],
 [75, 1],
 [78, 1],
 [80, 1],
 [82, 1],
 [85, 1],
 [87, 1],
 [90, 1],
 [92, 1],
 [95, 1],
 [97, 1],
 [99, 1]
])

X = data[:, 0].reshape(-1, 1)
y = data[:, 1].reshape(-1, 1)
m = len(y)

X = (X - np.mean(X)) / np.std(X)

def sigmoid(z):
    return 1 / (1 + np.exp(-z))

theta0 = 0
theta1 = 0
alpha = 0.1
iterations = 1000
costs = []

for i in range(iterations):
    z = theta0 + theta1 * X
    h = sigmoid(z)
    cost = -(1/m) * np.sum(y * np.log(h + 1e-8) + (1 - y) * np.log(1 - h + 1e-8))
    costs.append(cost)
    theta0 -= alpha * (1/m) * np.sum(h - y)
    theta1 -= alpha * (1/m) * np.sum((h - y) * X)

print(theta0, theta1, costs[-1])

plt.plot(range(iterations), costs)
plt.xlabel('Iterations')
plt.ylabel('Cost')
plt.title('Cost Function')
plt.show()

plt.scatter(X, y, color='red')
x_vals = np.linspace(min(X), max(X), 100)
y_pred = sigmoid(theta0 + theta1 * x_vals)
plt.plot(x_vals, y_pred, color='blue')
plt.show()


#performance metrics
y_pred = (sigmoid(theta0 + theta1 * X) >= 0.5).astype(int)

TP = np.sum((y_pred == 1) & (y == 1))
TN = np.sum((y_pred == 0) & (y == 0))
FP = np.sum((y_pred == 1) & (y == 0))
FN = np.sum((y_pred == 0) & (y == 1))

accuracy  = (TP + TN) / (TP + TN + FP + FN)
precision = TP / (TP + FP + 1e-8)
recall    = TP / (TP + FN + 1e-8)
f1        = 2 * (precision * recall) / (precision + recall + 1e-8)

print("Accuracy :", accuracy)
print("Precision:", precision)
print("Recall   :", recall)
print("F1-score :", f1)


```
