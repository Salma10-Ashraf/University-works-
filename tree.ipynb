```python
import pandas as pd    #1
```

```python
data=pd.read_csv('iris.csv')   #2
data.head()
```

```python
x=data.drop('Species',axis=1)  #3
x.head()
```

```python
y=data['Species']
y.head()
```

```python
y.value_counts()   #4
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,train_size=0.8)  #5
```

```python
x_test.head()
```

```python
x_train.head()
```

```python
from sklearn import tree
```

```python
from sklearn.tree import DecisionTreeClassifier  #6
```

```python
tree_model= DecisionTreeClassifier(criterion ='entropy')
```

```python
tree_model.fit(x_train,y_train)    #7
```

```python
from sklearn.metrics import accuracy_score
```

```python
train_pred=tree_model.predict(x_train)
train_acc=accuracy_score(y_train,train_pred)
print('tree training accraccy = ',train_acc)
```

```python
test_pred=tree_model.predict(x_test)
test_acc=accuracy_score(y_test,test_pred)    #10
print('tree testing accraccy = ',test_acc)
```

```python
import matplotlib.pyplot as plt
```

```python
tree.plot_tree(tree_model,feature_names=x.columns, filled=True, class_names=['setosa','versicolor','virginica'])
plt.show()
```
