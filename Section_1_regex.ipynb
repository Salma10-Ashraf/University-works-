```python
import re
```

***findall()*** \
function returns a list containing all matches.

```python
txt = "The rain in Spain"
x = re.findall("ai", txt)
print(x)
x = re.findall("Portugal", txt)
print(x)
```

***search()*** \
function searches the string for a match, and returns a Match object if there is a match.

If there is more than one match, only the first occurrence of the match will be returned:

```python
x = re.search(r"\s", txt)

print("The first white-space character is located in position:", x.start())

x = re.search("Portugal", txt)

print(x)
```

***`split() `*** \
function returns a list where the string has been split at each match:

```python
x = re.split(r"\s", txt)
print(x)

# Limit split to the first occurrence of whitespace to isolate the first word/token
x = re.split(r"\s", txt, 1)
print(x)
# Split at the first 2 whitespace occurrences, resulting in a 3-item list
x = re.split(r"\s", txt, 2)

print(x)

# Split at the first 3 whitespace occurrences, resulting in a 4-item list
x = re.split(r"\s", txt, 3)
print(x)
```

***sub()*** \
function replaces the matches with the text of your choice:



```python
x = re.sub(r"\s", "9", txt)
print(x)
x = re.sub(r"\s", "9", txt, 2) #first 2 matches
print(x)
```

```python
text_try = """Data analytics is a field that involves 123 different techniques. It helps us understand complex patterns.
This text includes various characters like !, @, #, $, and %.
There are also some lowercase vowels here: a, e, i, o, u.
We can find more digits such as 7, 8, 9, 0, 1. Another line starts here.
Let's see if we can find "Data" at the beginning again.
This is the end of the line
And this is another line."""
#Finding Digits
x = re.findall(r"\d", text_try) #Perl Notation
print(x)


x = re.findall(r"[0-9]", text_try) #Minimal Notation
print(x)
```

```python
#Finding Non-Digits
x = re.findall(r"\D", text_try) #Perl Notation
print(x)

x = re.findall(r"[^0-9]", text_try) #Minimal Notation
print(x)
```

```python
#Word Starts with ^
x = re.findall(r"^Data", text_try)
print(x)

#Word Ends with $
x = re.findall(r"line\.$", text_try,re.MULTILINE)
print(x)

#list of Chars
x = re.findall(r"[aeiou]", text_try)
print(x)
```

```python
text_cases = """This is a sample text for regex testing. Alpha numeric. abcdefgh. The lazy brown fox jumps over the quick dog.

Here are some words ending in b: grab, climb, lamb. This is not a grottos but a grotto. And the raven flew by. Another raven flew by the grotto.

We need to find repeated words like hello hello or the the. This is a word word sequence. Not a bug bug, but a bug.

Pattern for 'bab' type strings: bab, bbabb, bbbabbb. aba is not a match. abb is not a match.

1 This is the first line. word
2 Another line here second
10 Third line for testing testword
1000 This is a long line that ends with word
1000 This is a long line that ends with word 1000
First, word with punctuation! Second, word of sentence. Three, words here?

"""
```

```python
re.findall(r"[a-zA-Z]+", text_cases) #words with at least 1 chracter in it
```

```python
re.findall(r"b[a-zA-Z]+", text_cases) #words starts with [b] and have at least 1 letter other than the {b}
```

```python
re.findall(r"\b(\w+)\s+\1\b", text_cases) #repated words in row
```

```python
re.findall(r"\b(\w+)\s\1\b", text_cases)
```

```python
re.findall(r"[a-zA-Z]*bab[a-zA-Z]*", text_cases) #words have bab in it
```

```python
pattern = r"^[0-9]+.*[a-zA-Z]$"
x = re.findall(pattern, text_cases ,re.MULTILINE) #lines start with a number and ends with a word
x
```

```python
sentences = [
    "Hello, world!",
    "Data is powerful.",
    "'Wait,' she said.", # Starts with a quote
    "123 numbers don't count."
]

# Pattern: Starts with optional non-word chars (like quotes),
# then captures the first block of letters.
pattern = r"^[^\w]*([A-Za-z]+)"
for sentence in sentences:
  print(re.findall(pattern,sentence))
print(r"[^\w]*([A-Za-z]+) matches anything other than special characters and discard them then group the word after")
```

```python
import re

text = ".Wait"
pattern = r"^[^\w]*([A-Za-z]+)"

match = re.search(pattern, text)

if match:
    print(f"Full Match (group 0): '{match.group(0)}'")
    print(f"Capture Group (group 1): '{match.group(1)}'")
```

```python
text_with_binary = """Some binary examples: 01101, 011010, , 010.
Also, 10101010 and 11110000 are sequences of 0s and 1s that could be interpreted as binary.
"""
re.findall(r"\b[0-1]+\b",text_with_binary,re.MULTILINE) #detect binary numbers
```

```python
text_with_hex_colors = """Some common CSS hex color include #FF0000 (red),
#00FF00 (green), and #0000FF (blue). You might also find shorthand colour like #FFF (white)
or #333 (dark gray). Other examples include #ABCDEF and #1a2b3c."""

re.findall(r"#[0-9a-fA-F]{6}|#[0-9a-fA-F]{3}",text_with_hex_colors,re.MULTILINE) #detect hex colors
```

```python
print(re.findall(r"\b(colo?.r)\b",text_with_hex_colors)) #detect color in both ways color or colour
print(r"?. this means it may have a letter here or not")
print(re.findall(r"\b(colour|color)\b",text_with_hex_colors)) #detect color in both ways color or colour
```

```python
test_string = "The the cat sat on the mat. Contact us at info@nlp.edu or (123) 456-7890 or at (555) 879-9800"
pattern = r"\([0-9]{3}\) [0-9]{3}-[0-9]{4}" #detects the US written phone numbers
re.findall(pattern,test_string)
```

```python
text_with_twitter_handles = """Connect with us on Twitter! Follow @ColabResearch, @GoogleAI, and @Python_Devs. You can also find @news on the platform. Remember that @verylongusername is not a valid handle as it exceeds 15 characters, but @short_handle is fine."""
pattern = r"@[a-zA-Z0-9_]{1,15}"#detects twitter handles
re.findall(pattern,text_with_twitter_handles)
```

```python
haha_txt = "haha hahaha hahahaha hahahahab"
pattern = r"\b(?:ha){2,}\b"
re.findall(pattern,haha_txt) #any variation of "haha"
```

```python
pattern = r"\$[0-9]+(?:\.[0-9]{2,})?"
text_with_dollar_amounts = """The price of the item is $19.99. or $19-99 We also have items for $5, $100.50, and a premium product at $1250.75. Some amounts are simply $0.50 or $20. There are also prices like $1.234 which might not be valid currency formats."""
re.findall(pattern,text_with_dollar_amounts) #usd currency
```

```python
pattern = r"\b(?:0[1-9]|[12][0-9]|3[01])/(?:0[1-9]|1[0-2])/[0-9]{4}\b"
text_with_dates = """Some important dates include 25/12/2023 (Christmas), 01/01/2024 (New Year's Day), and 14/02/2023 (Valentine's Day). Also consider 31/03/2023 and 05/09/2024."""
re.findall(pattern,text_with_dates)#Dates
```
