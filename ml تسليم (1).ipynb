```python
import pandas as pd
```

```python
data=pd.read_csv('car data.csv')
data.head()
```

```python
from sklearn.preprocessing import LabelEncoder

encoder = LabelEncoder()

data['Fuel_Type'] = encoder.fit_transform(data['Fuel_Type'])
data['Seller_Type'] = encoder.fit_transform(data['Seller_Type'])
data['Transmission'] = encoder.fit_transform(data['Transmission'])
data['Car_Name'] = encoder.fit_transform(data['Car_Name'])
```

```python
print(data.dtypes)
```

```python
x=data.drop('Selling_Price',axis=1)
x
```

```python
y=data['Selling_Price']
y
```

```python
y.value_counts()
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,train_size=0.8)
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error,mean_absolute_error,root_mean_squared_error
```

```python
lr=LinearRegression()
```

```python
lr.fit(x_train,y_train)
```

```python
train_pred=lr.predict(x_train)
train_mse=mean_squared_error(y_train,train_pred)
print('mean_squared_error on training=',train_mse)
```

```python
test_pred=lr.predict(x_test)
test_mse=mean_squared_error(y_test,test_pred)
print('mean_squared_error on testing=',test_mse)
```

```python
mae=mean_absolute_error(y_test,test_pred)
```

```python
print('mean_absolute_error=',mae)
```

```python
rmse=root_mean_squared_error(y_test,test_pred)
```

```python
print('root_mean_squared_error=',rmse)
```

```python
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import accuracy_score,mean_squared_error,mean_absolute_error,r2_score
```

```python
rf = RandomForestRegressor()
```

```python
rf.fit(x_train,y_train)
```

```python
y_pred = rf.predict(x_test)
mae = mean_absolute_error(y_test,y_pred)
print("MAE:",mae)
```

```python
mse = mean_squared_error(y_test,y_pred)
print("MSE:",mse)
```

```python
r2 = r2_score(y_test,y_pred)
print("R2 Score:",r2)
```
